# Data Cleaning and Exploratotory Analysis

```python
#Importation of libraries
import pandas as pd
import numpy as np
import yfinance as yf
from dateutil import parser
```

```python
#Read data
df = pd.read_csv("Data.csv")
df
```

## Data Cleaning

```python
df = df.replace(to_replace='^.*<.*$', value=np.nan, regex=True)
df['MarketingChannel_clean'] = df['Marketing Channel'].str.strip().replace('Paid', 'Paid Search', regex = True)
```

```python
df['MarketingChannel_clean'] = df['Marketing Channel'].str.strip().replace('Organic ', 'Organic Search', regex = True)
```

```python
df['MarketingChannel_clean'] = df['Marketing Channel'].str.strip().replace('Email ', 'Email Marketing', regex = True)
```

```python
df['MarketingChannel_clean'] = df['Marketing Channel'].str.strip().replace('Social Media - Facebook', 'Facebook', regex = True)
df['MarketingChannel_clean'] = df['Marketing Channel'].str.strip().replace('Social Media-Facebook', 'Facebook', regex = True)
df['MarketingChannel_clean'] = df['Marketing Channel'].str.strip().replace('Social Media - TikTok', 'TikTok', regex = True)
df['MarketingChannel_clean'] = df['Marketing Channel'].str.strip().replace('Social Media - Pinterest', 'Pinterest', regex = True)
df['MarketingChannel_clean'] = df['Marketing Channel'].str.strip().replace('Social Media - Instagram', 'Instagram', regex = True)
```

```python
df['Product_Name_Clean'] = df['Product_Name'].str.encode('utf-8').str.decode('ascii', 'ignore')
```

```python
df['Region_clean'] = df['Region'].str.strip().fillna( 'Unknown')
df['AccountCreationMethodclean'] = df['AccountCreationMethod'].str.strip().fillna('Unknown')
```

```python
country_to_currency = {
    "NL": "EUR",
    "IL": "ILS",
    "SG": "SGD",
    "GB": "GBP",
    "US": "USD",
    "JP": "JPY",
    "NZ": "NZD",
    "FR": "EUR",
    "DE": "EUR",
    "MX": "MXN",
    "IN": "INR",
    "ES": "EUR",
    "AE": "AED",
    "IT": "EUR",
    "CA": "CAD",
    "AR": "ARS",
    "AU": "AUD",
    "BR": "BRL",
    "SA": "SAR",
    "ZA": "ZAR"
}
df["Currency_clean"] = df["CountryCode"].map(country_to_currency)
df["Currency_clean"] = df["Currency"].fillna(
    df["CountryCode"].map(country_to_currency)
)
df["Currency_clean"] = df["Currency_clean"].fillna(
    "Unknown")
```

```python
df["CountryCode_clean"] = df["CountryCode"].fillna(
    "Unknown")
```

```python
df["ProductID_clean"] = df["ProductID"].fillna(
    "Unknown")
```

```python
df["Currency_extracted"] = df["Price_"].str.extract(r"([A-Z]{3})")

df["Price_numeric_clean"] = (
    df["Price_"]
        .str.extract(r"([\d,\.]+)")[0]
        .str.replace(",", "", regex=False)
        .astype(float)
)



unique_currencies = df["Currency_extracted"].dropna().unique()

rates = {}

df["Currency_extracted"] = df["Currency_extracted"].fillna(df["Currency_clean"])

for cur in unique_currencies:
    if cur == "USD":
        rates[cur] = 1
    else:
        ticker = f"{cur}USD=X"
        data = yf.Ticker(ticker).history(period="1d")
        rates[cur] = data["Close"].iloc[-1]

df["Price_USD"] = (
    df["Price_numeric_clean"] *
    df["Currency_extracted"].map(rates)
)
```

```python

def parse_date(val):
    if pd.isna(val) or str(val).strip() == "":
        return None
    s = str(val).strip().replace(".", "-")
    try:
        return parser.parse(s).strftime("%m/%d/%Y")
    except:
        return None

date_cols = ["Purchase Ts", "Created On", "Ship Ts", "Delivery Ts", "Refund Ts"]

for col in date_cols:
    df[col + "_clean"] = df[col].apply(parse_date)
```

```python
#Save to new file
df.to_csv("Orderfilecleandata.csv", index=False)
print("Done!")
```
